# Hospital Analytics - Exploratory Data Analysis

Quick EDA on the cleaned dataset produced by `src/transform.py`.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/cleaned/hospital_cleaned.csv', parse_dates=['Date of Admission', 'Discharge Date'])
df.head()
```

```python
df.info()
df.describe(include='all')
```

## Revenue by Department

```python
dept_revenue = df.groupby('Department')['Billing Amount'].sum().sort_values(ascending=False)
dept_revenue.plot(kind='bar', figsize=(10,5), title='Total Revenue by Department')
plt.ylabel('Revenue ($)')
plt.tight_layout()
plt.show()
```

## Monthly Admissions Trend

```python
monthly = df.groupby(['Admission Year', 'Admission Month']).size()
monthly.plot(kind='line', figsize=(12,5), title='Monthly Admissions Trend', marker='o')
plt.ylabel('Admissions')
plt.tight_layout()
plt.show()
```

## Top Doctors by Patient Count

```python
top_doctors = df['Doctor'].value_counts().head(10)
top_doctors.plot(kind='barh', figsize=(8,6), title='Top 10 Doctors by Patient Count')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## Age Group & Gender Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(14,5))
df['Age Group'].value_counts().plot(kind='bar', ax=axes[0], title='Age Group Distribution')
df['Gender'].value_counts().plot(kind='pie', ax=axes[1], autopct='%1.1f%%', title='Gender Distribution')
plt.tight_layout()
plt.show()
```

## Insurance Provider Breakdown

```python
df['Insurance Provider'].value_counts().plot(kind='bar', figsize=(8,5), title='Patients by Insurance Provider')
plt.tight_layout()
plt.show()
```
